```python
import pandas as pd
import os
from sqlalchemy import create_engine

engine = create_engine('sqlite:///inventory2.db', echo=False)

print("Connected Successfully")
```

```python

def ingest_db(df, table_name, engine):
    df.to_sql(table_name, con= engine, if_exists ='replace', index=False)


for file in os. listdir('data'):
    if '.csv' in file:
        df = pd.read_csv('data/'+file)
        print(df.shape)
        ingest_db(df, file[:-4],engine)
```

```python

import duckdb
import os
import logging
import time

logging.basicConfig(filename="logs/ingestion_db.log", filename=logging.DEBUG,format="%(asctime)s- %(levelname)s- %(message)s",
filemode="a")

con = duckdb.connect("inventory.db")

def load_raw_data():
    '''this function will load the CSVs as dataframe and ingest into db'''

    start = time.time()
    for file in os.listdir("data"):

     if file.endswith(".csv"): 

        table_name = file[:-4]

        file_path = os.path.join("data", file)

        print(f"Processing {file}")
        logging.info(f'Ingesting {file} in db')

        con.execute(f"""
            CREATE OR REPLACE TABLE {table_name} AS
            SELECT *
            FROM read_csv_auto('{file_path}')
        """)

        print(f"{file} completed")
    end = time.time()
    total_time = (end - start)/60
    logging.info("---------------Ingestion Complete-----------")
    logging.info(f'\nTotal Time Taken: {total_time} minutes')

if __name__ == '__main__':
   load_raw_data()   
```
